```python
# 7.	Write a program find the maximum number from the given input list
Input_list = [99,82, 62, 61, 54, 71, 89, 75, 73]
max = 0
for i in range(len(Input_list)):
    for j in range(i+1, len(Input_list)):
        if (Input_list[i] > Input_list[j]) and (Input_list[i] > max):
            max = Input_list[i]
        elif (Input_list[i] < Input_list[j]) and (max < Input_list[j] ):
            max = Input_list[j]
        else:
            max = max
print(max)
```

```python
# 18.	Write a program create a random list of length 10 and print all the elements except the elements which are divisible by 4. 
import random

random_number = []
final_list = []
for _ in range(10):
    random_number.append(random.randint(1,100))
        
print(random_number)
for i in range(10):
    if random_number[i]%4==0:
        final_list.append(random_number[i])
print(f"random_number divisable by 4 {final_list}")
```
