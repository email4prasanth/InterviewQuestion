```python
# WAP to read temperature in centigrade and display a suitable message according to temperature state below
# Temp < 0 → Freezing weather
# Temp 0-10 → Very Cold weather
# Temp 10-20 → Cold weather
# Temp 20-30 → Normal in Temp
# Temp 30-40 → Its Hot
# Temp >= 40 → Its Very Hot

temp = float(input("Enter the temparature in centigrade: "))

def weather(temp):
    try:
        if temp <0 :
            return "Freezing weather"
        elif 0 <= temp <10:
            return "Very Cold weather"
        elif 10<temp<20:
            return "Cold weather"
        elif 20<temp<30:
            return "normal in Temp"
        elif 30<temp<40:
            return "Its Hot"
        else:
            return "Very Hot"
    except Exception as e:
        raise Exception(f"Error in main file {e}")
def main():
    con = weather(temp)
    print(con)
if __name__ == "__main__":
    main()
```

```python
# 17. WAP to check whether a triangle is Equilateral, Isosceles or Scalene.
#
# Triangle Types:
#
# Equilateral Triangle:
# All three sides are equal
#
# Isosceles Triangle:
# Any two sides are equal
#
# Scalene Triangle:
# All three sides are different
x = int(input("Enter x value:"))
y = int(input("Enter y value:"))
z = int(input("Enter z value:"))
def tri_type(x,y,z):
    if (x==y==z):
        return "Equilateral Triangle"
    elif (x==y and x!=z) or (x==z and x!=y) or (y==z and y!=x):
        return "Isosceles Triangle"
    else:
        return "Scalene Triangle"
def main():
    triangle_type = tri_type(x,y,z)
    print(triangle_type)
if __name__ == "__main__":
    main()
```

```python
# 20. WAP to accept a grade and declare the equivalent description.
#
# Grade Description:
#
# E -> Excellent
# V -> Very Good
# G -> Good
# A -> Average
# F -> Fail
#
# The program should accept a grade character from the user
# and display its equivalent description using conditional statements.
grade = input("Enter Grade")
def val(grade):
    if grade == 'E':
        return "Excellent"
    elif grade == 'V':
        return "Very Good"
    elif grade == 'G':
        return "Good"
    elif grade == 'A':
        return "Average"
    elif grade == 'F':
        return "Fail"
    else:
        return "Invalid Grade"
def main():
    print(val(grade))

if __name__ == "__main__":
    main()
```
